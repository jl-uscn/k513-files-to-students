## 1.5 · Screening, not concluding

A pairplot is a **shortlist of pairs of continuous variables worth drawing scatterplots**, and a heatmap below uses the same shortlist to show
correlation coefficients. Neither is a chart you would put in a report.

`corr()` is numeric-only, so select **only continuous** variables you want into the list.


```python
health_cols = ['Life Expectancy Female', 'Infant Mortality Rate',
                'Health Exp/Capita', 'Population Urban %']

sns.pairplot(world_df[health_cols])
plt.show()
```

```python
sns.heatmap(world_df[health_cols].corr(), annot=True)
plt.show()
```

Do not read strength off that heatmap yet, and do not put a threshold on it — Week 3 is
where correlation gets read properly. For now it does one job: it tells you **which pair to draw
next**.

## 2.2 · The foundation's own tables

Three small tables, built here directly. Read them — they are short enough to read and understand, and Part 2 is entirely about what happens to their **shapes**.


```python
office_df = pd.DataFrame({
    'office_id': [101, 102, 103, 104, 105, 106, 107, 108, 109],
    'city':      ['Nairobi', 'Lagos', 'Dhaka', 'Lima', 'Amman', 'Manila',
                  'Bogota', 'Suva', 'Rabat'],
    'region':    ['Africa', 'Africa', 'Asia', 'The Americas', 'Middle East', 'Asia',
                  'The Americas', 'Oceania', 'Africa'],
    'staff':     [24, 31, 18, 12, 9, 27, 15, 6, 11],
})
office_df
```

```python
grants_h1 = pd.DataFrame({
    'grant_id':  [9001, 9002, 9003, 9004, 9005, 9006, 9007],
    'office_id': [101, 101, 102, 103, 104, 105, 101],
    'amount':    [45000, 22000, 61000, 18000, 30000, 12500, 27000],
})
display(grants_h1)

grants_h2 = pd.DataFrame({
    'grant_id':  [9008, 9009, 9010, 9011, 9012],
    'office_id': [102, 103, 104, 106, 107],
    'amount':    [38000, 24000, 16000, 52000, 9500],
})
display(grants_h2)

print(grants_h1.shape, grants_h2.shape)
```

## 2.3 · Combine, part one: more rows of the same kind

Two half-year files, the same columns in both. This is `concat()`, not `merge()` — you are adding
rows, not adding facts about rows you already have.

Make `ignore_index=True` a reflex. Without it you get two rows numbered 0, two numbered 1, and a
bug three cells later. Also note that both grants_h1 and grants_h2 have the same number of columns and column names. That's important because `pd.concat` knows how to append them.


```python
grant_df = pd.concat([grants_h1, grants_h2], ignore_index=True)
print(grant_df.shape)
grant_df
```

## 2.4 · Combine, part two: more facts about the rows you have

Now the join. Before you write it: **9 offices, 12 grants.** Think why number of grants is higher than that of office. Predict the number of rows an inner
merge returns, write your prediction down, and then run it.

**Canvas questions 3 and 4** ask for the two row counts.


```python
print('offices :', office_df.shape)
print('grants  :', grant_df.shape)

inner_df = office_df.merge(grant_df, on='office_id', how='inner')
print('inner   :', inner_df.shape)
inner_df
```

Row count is not the whole story, and this is the cell that shows why. **Canvas question 5**

```python
print('offices in  :', office_df['office_id'].nunique())
print('offices out :', inner_df['office_id'].nunique())
```

## 2.5 · Reshape

The foundation's finance team keeps quarterly spending like this — one row per office, one column
per quarter. It is a good layout for a person to read and the wrong one if you want to compare regions in different quarters, which needs tidy data.

**Canvas question 2** asks how many rows come out of the melt.


```python
spend_df = pd.DataFrame({
    'city': ['Nairobi', 'Lagos', 'Dhaka', 'Lima', 'Manila'],
    'Q1':   [82000, 91000, 47000, 38000, 66000],
    'Q2':   [79000, 88000, 52000, 41000, 71000],
    'Q3':   [95000, 84000, 49000, 36000, 74000],
    'Q4':   [88000, 97000, 55000, 44000, 69000],
})
spend_df
```

One argument decides the whole thing: **which column is not a quarter?** That is what stays
put while the rest get folded down.


```python
long_df = spend_df.melt(id_vars='____', var_name='quarter', value_name='spend')
print(spend_df.shape, '->', long_df.shape)
long_df.head()
```

Now it draws. This is the whole reason to melt — the chart on the wide (untidy) table cannot be
created at all, because there is no `quarter` column to pass to `x=`.


```python
plt.figure(figsize=(10, 4))
sns.barplot(data=long_df, x='quarter', y='spend', hue='city', errorbar=None)
plt.show()
```

# Part 3 · A chart that is true and misleading

Two charts of the same column. Both are correctly drawn. They give a reader different answers to the
question *"which regions have the big populations?"*


```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

sns.barplot(data=world_df, x='Region', y='Population (M)', order=REGION_ORDER,
            errorbar=None, ax=axes[0])
axes[0].set_title('mean population in regions')
axes[0].tick_params(axis='x', rotation=30)

sns.barplot(data=world_df, x='Region', y='Population (M)', order=REGION_ORDER,
            estimator='median', errorbar=None, ax=axes[1])
axes[1].set_title('median population in regions')
axes[1].tick_params(axis='x', rotation=30)

sns.boxplot(data=world_df, x='Region', y='Population (M)', order=REGION_ORDER, ax=axes[2])
axes[2].set_title('distribution of population per country')
axes[2].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

The boxplot is unreadable, and that is a finding rather than a mistake. Redraw the boxplot without the four largest countries in the world so you can see the boxes — then remember that you have removed real countries to make a picture legible, and that this chart is a diagnostic, not something you would show anybody without telling the removal first.


```python
smaller_df = world_df.sort_values(by='Population (M)', ascending=False).tail(204)

plt.figure(figsize=(10, 4))
sns.boxplot(data=smaller_df, x='Region', y='Population (M)', order=REGION_ORDER)
plt.show()
```
