## 1.3 · The heatmap

`sns.heatmap()` draws the same table with color instead of rows. Three blanks:

- **`annot=`** decides whether the numbers are printed in the squares. Without it you get color and
  nothing else, which is pretty and useless.
- **`cmap=`** picks the colors. `'vlag'` runs blue through white to red.

For a correlation matrix the last two matter more than they look. A correlation runs from −1 to +1
and the **sign** is half the meaning, so you want a color scale with a middle — one where zero is
pale, strong positive is one color and strong negative is the other. A scale without a middle makes
−0.7 and +0.7 look like completely unrelated values.

```python
plt.figure(figsize=(11, 9))
sns.heatmap(data=corr_df, annot=____, fmt='.2f', cmap='vlag')
plt.title('Boston housing — correlation of every pair')
plt.show()
```

**`fmt='.2f'` is not decoration.** Without it the printed numbers come out at different
widths — `-0.0074` next to `1` next to `-0.12` — and in a tight row they run into each other.
Two decimal places gives every square the same width.

## 1.4 · The pairplot

The heatmap gives you a number for every pair. It does not give you a **picture** for any of them —
and today's session was about exactly the thing a number cannot see.

`sns.pairplot()` draws a scatterplot for every pair, with each column's own histogram down the
diagonal.

**The blank is the `vars=` list, and it is the whole decision.** Left to itself, `pairplot` draws
every column against every other one: 13 columns is 169 panels, which takes a while to render and
is unreadable when it arrives. Naming five columns gives you 25 panels you can actually look at.

Put `MEDV` in the list — it is the target — plus the two columns from question (a) and the two from
