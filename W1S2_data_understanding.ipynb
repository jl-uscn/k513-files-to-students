`.info()` is the one to run on every new dataset, and it answers three questions at once.

```python
boston_df.info()
```

Read it in this order:

1. **How many rows.** `RangeIndex: 506 entries` — that should match `.shape`.
2. **The column names, and how they are spelled.** Getting a column name wrong is the single most
   common error you will hit in the next three weeks.
3. **The non-null count.** Every column here says `506 non-null`, which means nothing is missing.
   That is unusual. Real data is not this tidy, and a column showing fewer non-nulls than there are
   rows is telling you something.

`float64` means numbers with decimals and `int64` means whole numbers. A column of text shows up as
`object` (or `str`, on newer versions of pandas) — none of them do here, but two of them will in a
moment.

## 3 · Reading one variable

Four questions, asked of one column at a time. **Centre · Spread · Shape · Outliers.**

`.describe()` answers all four at once for every numeric column. Running it is one word; reading it
is the skill.

```python
boston_df.describe()
```

Look at **`CRIM`**, the per-capita crime rate, and read down its column.

| | |
|---|---|
| `mean` | 3.614 |
| `50%` (the median) | 0.257 |
| `max` | 88.976 |

The mean is **fourteen times** the median, and the maximum is three hundred times it. Almost every
tract in this dataset has very little crime, and a handful have a great deal.

That is what a **right-skewed** variable looks like in numbers, and it is the fastest read available
on a new dataset: compare `mean` to `50%`. If the mean sits well above the median, a few large
values are pulling it up, and the mean is not a typical value for anything.

Now ask yourself which of those two numbers you would put in a report about crime in Boston. There
is no comfortable answer, and that is the point.

Anything `.describe()` does not give you, call directly — on the whole table, or on one column.

```python
boston_df.skew()      # roughly -1 to 1 is symmetric; below -1 left-skewed; above 1 right-skewed
```

```python
print("Skewness of MEDV:  ", boston_df['MEDV'].skew())
print("10th percentile:   ", boston_df['MEDV'].quantile(q=0.1))
print("90th percentile:   ", boston_df['MEDV'].quantile(q=0.9))
print("3 largest values:  ", boston_df['MEDV'].nlargest(3).tolist())
print("3 smallest values: ", boston_df['MEDV'].nsmallest(3).tolist())
```

Note the square brackets: `boston_df['MEDV']` pulls out one column. Spelling and capitalization have
to match exactly — `MEDV` works, `medv` does not.

## 4 · Not every variable is a number

Back to the question you left hanging in Now You Try 1.

*(If you did not get `bike_df` loaded in section 2, run the cell below and carry on — nothing here
depends on having done it yourself.)*

```python
bike_df = pd.read_csv(BIKE_URL)     # safety net, in case section 2 did not finish
```

`.describe()` dropped `date` and `weather` because it had nothing numeric to summarise. You can ask
for those columns specifically.

```python
bike_df.describe(exclude='number')
```

`count` is how many rows have a value, `unique` is how many *distinct* values there are, `top` is
the most common one and `freq` is how often it occurs.

`weather` has 3 unique values across 731 days. `date` has 731 — one per row, which tells you it is
an identifier rather than something to analyse.

(`exclude='number'` is the readable way to say "everything that is not numeric". `include='object'` or `include='O'`
does the same job and you will see it in older material.)

To see the actual categories and their counts, use `value_counts()`.

```python
bike_df['weather'].value_counts()
```

```python
bike_df['weather'].value_counts(normalize=True)   # the same thing as shares of the total
```

This is the count half of a pivot table — the thing you already do in Excel — in one line.

It is also how you find out that a column you thought had four categories actually has forty-seven,
six of them the same category spelled differently. That is not a hypothetical.

**And it works just as well on numeric columns**, which is the important part. `CHAS` in Boston
Housing is stored as a number:

```python
boston_df['CHAS'].value_counts()
```

Two rows. `CHAS` is not a continuous variable at all — it is 1 if the tract touches the Charles
River and 0 if it does not. A category, wearing a number's clothes.

pandas cannot tell the difference. It saw a number, assumed continuous, and reported a mean and a
standard deviation for it in `.describe()` without complaint.

(For a 0/1 column the mean is at least interpretable — it is the *proportion* of tracts on the
river, about 0.07. For a category coded 1, 2, 3 the mean means nothing at all.)

Deciding what each column really is, is your job. It has to be done before Week 4, and nothing will
warn you if you get it wrong.

### Stretch — only if you finished early

Two more one-liners. Both are useful; neither is hard.

**`corr()`** returns a table of every pair of numeric columns. Each number is *r*, the strength of
the linear (straight-line) relationship between two variables, running from −1 to +1. Today you only need to
be able to get it and spot the large values. Reading it properly is Week 3.

```python
boston_df.corr()
```

`corr()` needs every column to be numeric. `bike_df` has text columns, so take a subset first — a
list of column names inside a second set of square brackets.

```python
bike_df[['temp in Celsius', 'humidity', 'windspeed', 'num_shared']].corr()
```

Ignore the diagonal — every variable correlates perfectly with itself. Of the remaining pairs, which
two variables have the strongest relationship, and does its direction make sense to you?

(Note the column name `temp in Celsius`, spaces and all. Real column names are ugly. You copy them
exactly as `.info()` shows them.)

**`sort_values()`** sorts. Descending is the fast way to look at the extreme end of a variable — the
tracts driving that `CRIM` mean you were suspicious about.

```python
boston_df.sort_values(by='CRIM', ascending=False).head(10)
```

```python
boston_df.sort_values(by=['CHAS', 'CRIM'], ascending=False).head(10)   # more than one column: use a list
```
