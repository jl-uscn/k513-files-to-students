## 1 · Loading the data

Python on its own does not know what a spreadsheet is. We borrow that ability from a library called **pandas**.

Run the cell below. Nothing visible will happen — that is correct. You have just told Python "go and get pandas, and let me call it `pd` for short".


```python
import pandas as pd
```

Now we point pandas at a data file. It lives at a fixed web address, so there is nothing for you to download and no folder for you to find.

`pd.read_csv(...)` reads the file and hands back a table. We store that table in a name of our choosing — `df`, short for **d**ata**f**rame, which is what pandas calls a table.

Again, running this shows nothing. Loading is silent.


```python
DATA_URL = "https://raw.githubusercontent.com/jl-uscn/k513-data/main/BostonHousing.csv"

df = pd.read_csv(DATA_URL)
```

Now ask Python to show you `df`. Type the name of a thing on the last line of a cell and Colab prints it.


```python
df
```

There it is: **506 rows and 13 columns**.

Each row is one neighborhood in the Boston area. Each column is one thing that was measured about it.

Two words we will use all term:

| Everyday word | What we will call it |
|---|---|
| a row | an **observation** (or a case, or a data point) |
| a column | a **variable** (or, once we start modeling, a **feature**) |


### The columns

| Column | Meaning |
|---|---|
| `CRIM` | per-capita crime rate |
| `ZN` | % of residential land zoned for large lots |
| `INDUS` | % of land that is non-retail business |
| `CHAS` | 1 if the tract touches the Charles River, 0 if not |
| `NOX` | nitric-oxide concentration (air quality) |
| `RM` | average number of rooms per home |
| `AGE` | % of homes built before 1940 |
| `DIS` | distance to Boston employment centers |
| `RAD` | accessibility to radial highways |
| `TAX` | property-tax rate per \\$10,000 |
| `PTRATIO` | pupil-teacher ratio in local schools |
| `LSTAT` | % of the population classed as "lower status" |
| `MEDV` | **median home value, in \\$1,000s** |

`MEDV` is the one to remember. In Week 4 it becomes the thing we try to predict.

> A note on where this came from: the dataset was assembled in 1978, and some of its columns encode assumptions nobody would write down today. One column has already been removed from our copy. `LSTAT` is still here, and its definition is worth reading twice. Where data comes from is part of the analysis, not a footnote to it.

## 3 · One column at a time

To pull out a single column, put its name in square brackets and quotation marks.

Spelling and capitalization have to match exactly. `MEDV` works; `medv` does not.


```python
df['MEDV']
```

Once you have a column, you can ask it for a statistics.


```python
print(df['MEDV'].mean())
print(df['MEDV'].median())
print(df['MEDV'].min())
print(df['MEDV'].max())
```

Read those four numbers before you move on.

The mean is a little higher than the median. The maximum is a long way above both. That gap is not noise — it is telling you something about the shape of the data, and on Thursday we will give it a name.

(Remember: these are in thousands of dollars, in 1978.)

### All of it at once

`.describe()` runs those summaries on every numeric column simultaneously. It is usually the second thing you do to a new dataset, right after `.head()`.


```python
df.describe()
```

Scan the `mean` row against the `50%` row (that is the median).

Where they are close, the column is roughly symmetric. Where the mean sits well above the median — look at `CRIM` — a handful of extreme values are pulling it up. That single comparison is one of the fastest reads available on a new dataset, and you now know how to do it.


## 4 · Your first chart

A **histogram** answers: how are the values spread out?

It sorts the values into bins and draws a column for how many landed in each.


```python
df['MEDV'].plot(kind='hist', bins=30, title='Median home value ($1,000s)')
```

Look at the far right of that chart. There is a suspicious spike at the top end.

That is not a coincidence — values in the original data were capped at 50. Charts surface things that summary numbers hide, which is exactly why we draw them first and model second.
